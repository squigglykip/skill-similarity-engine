## 1. Audience Analysis

### Primary User: Future Skills Team
- **Role**: Strategic workforce analysts and planners
- **Needs**: Professional, data-backed reports to present transition options
- **Technical Level**: Moderate - understand methodology but need executive-ready outputs

### Secondary Audiences (Report Recipients):

#### 1. **Business Leaders & Division Heads**
- **Context**: Receiving reports about role transitions in their divisions
- **Needs**: Executive summary, impact assessment, clear recommendations
- **Language**: Strategic, business-focused, outcome-oriented

#### 2. **HR Business Partners**
- **Context**: Supporting employee transitions and workforce planning
- **Needs**: Practical guidance, timelines, development pathways
- **Language**: HR terminology, people-focused, actionable

#### 3. **Affected Colleagues/Employees**
- **Context**: Understanding their career options during role changes
- **Needs**: Personal relevance, clear pathways, skill development guidance
- **Language**: Supportive, opportunity-focused, accessible

#### 4. **Learning & Development Teams**
- **Context**: Designing training programs for transitions
- **Needs**: Skill gaps, development priorities, learning pathways
- **Language**: Educational, competency-focused, structured


## 2. Scenario Analysis

### Common Scenarios Requiring White Papers:

#### Scenario A: **Role Sunsetting**
- **Trigger**: Technology change, process automation, business model shift
- **Affected**: 5-50 people in similar roles
- **Urgency**: Medium-High (3-12 month timeline)
- **Story Focus**: "Where can these skills be applied?"

#### Scenario B: **Division Restructure**
- **Trigger**: Organisational change, merger, cost optimisation
- **Affected**: 20-200 people across multiple roles
- **Urgency**: High (1-6 month timeline)
- **Story Focus**: "How do we redeploy this talent?"

#### Scenario C: **Proactive Career Planning**
- **Trigger**: Future skills strategy, succession planning
- **Affected**: High-potential individuals or critical roles
- **Urgency**: Low-Medium (6-24 month timeline)
- **Story Focus**: "What are the growth opportunities?"

#### Scenario D: **Skills Gap Analysis**
- **Trigger**: New technology adoption, strategic capability building
- **Affected**: Teams needing to upskill or pivot
- **Urgency**: Medium (3-18 month timeline)
- **Story Focus**: "How do we build these capabilities?"

#### Scenario E: **Emergency Redeployment**
- **Trigger**: Market shock, regulatory change, crisis response
- **Affected**: Variable, often cross-functional
- **Urgency**: Very High (immediate to 3 months)
- **Story Focus**: "Who can adapt quickly?"


## 3. Data-Driven Story Types

### Story Architecture Based on Data Patterns:

#### **High Similarity Story** (Similarity > 0.7)
- **Narrative**: "Excellent transition opportunities available"
- **Focus**: Multiple viable pathways, low development effort
- **Tone**: Optimistic, opportunity-rich
- **Evidence**: Similarity scores, shared skills, position availability

#### **Medium Similarity Story** (Similarity 0.4-0.7)
- **Narrative**: "Strategic development opportunities with support"
- **Focus**: Targeted skill development, bridging pathways
- **Tone**: Realistic, development-oriented
- **Evidence**: Skill gaps, development timelines, success precedents

#### **Low Similarity Story** (Similarity < 0.4)
- **Narrative**: "Transformation opportunities requiring investment"
- **Focus**: Fundamental reskilling, longer-term transitions
- **Tone**: Honest but supportive, investment-focused
- **Evidence**: Transferable skills, alternative pathways, support programs

#### **Geographic Mobility Story**
- **Narrative**: "Expanded opportunities through location flexibility"
- **Focus**: Different locations offering better role matches
- **Evidence**: Position distribution, location-specific opportunities

#### **Cross-Division Story**
- **Narrative**: "Fresh perspectives through cross-functional moves"
- **Focus**: Applying skills in different business contexts
- **Evidence**: Cross-family similarity data, cultural fit indicators

## 5. Data Exploration for Story Generation


```python
# Get overview of available data for story generation
def explore_data_for_stories():
    """Explore the database to understand what stories we can tell."""
    
    print("=== DATA EXPLORATION FOR WHITE PAPER STORIES ===")
    
    # 1. Sample job roles for experimentation
    sample_jobs = pd.read_sql_query("""
        SELECT JobProfileID, JobProfile, JobFamily, JobFamilyGroup
        FROM jobs 
        ORDER BY JobProfile 
        LIMIT 10
    """, conn)
    
    print("\n📋 Sample Jobs for Testing:")
    print(sample_jobs.to_string(index=False))
    
    # 2. Similarity distribution analysis
    similarity_dist = pd.read_sql_query("""
        SELECT 
            CASE 
                WHEN similarity_score >= 0.7 THEN 'High (0.7+)'
                WHEN similarity_score >= 0.4 THEN 'Medium (0.4-0.7)'
                ELSE 'Low (<0.4)'
            END as similarity_category,
            COUNT(*) as count,
            ROUND(AVG(similarity_score), 3) as avg_score,
            ROUND(COUNT(*) * 100.0 / (SELECT COUNT(*) FROM job_similarities), 1) as percentage
        FROM job_similarities
        GROUP BY similarity_category
        ORDER BY avg_score DESC
    """, conn)
    
    print("\n📊 Similarity Distribution (Story Categories):")
    print(similarity_dist.to_string(index=False))
    
    # 3. Geographic distribution for mobility stories
    geographic_dist = pd.read_sql_query("""
        SELECT Location, COUNT(*) as position_count
        FROM positions 
        WHERE Location IS NOT NULL
        GROUP BY Location
        ORDER BY position_count DESC
        LIMIT 8
    """, conn)
    
    print("\n🌏 Geographic Distribution:")
    print(geographic_dist.to_string(index=False))
    
    return sample_jobs, similarity_dist, geographic_dist

sample_jobs, similarity_dist, geographic_dist = explore_data_for_stories()
```

## 6. Story Generation Logic - Branching Framework

This is the core of our white paper system - intelligent branching logic that determines what story to tell based on the data patterns we find.


```python
def analyze_role_for_story(job_profile_id):
    """
    Analyze a specific role to determine what story to tell.
    Returns data-driven insights for white paper generation.
    """
    
    print(f"\n=== STORY ANALYSIS FOR {job_profile_id} ===")
    
    # 1. Get job details
    job_details = pd.read_sql_query("""
        SELECT JobProfileID, JobProfile, JobFamily, JobFamilyGroup
        FROM jobs 
        WHERE JobProfileID = ?
    """, conn, params=[job_profile_id])
    
    if job_details.empty:
        print(f"❌ Job {job_profile_id} not found")
        return None
    
    job_info = job_details.iloc[0]
    print(f"📋 Job: {job_info['JobProfile']} ({job_info['JobFamily']})")
    
    # 2. Get career pathway opportunities
    pathways = pd.read_sql_query("""
        SELECT 
            cp.target_job_id,
            j.JobProfile,
            j.JobFamily,
            cp.similarity_score,
            cp.similarity_rank,
            cp.career_move_type,
            cp.shared_skills_count
        FROM career_pathways cp
        JOIN jobs j ON cp.target_job_id = j.JobProfileID
        WHERE cp.source_job_id = ?
        ORDER BY cp.similarity_rank
        LIMIT 10
    """, conn, params=[job_profile_id])
    
    print(f"\n🎯 Top Career Pathways ({len(pathways)} found):")
    if not pathways.empty:
        print(pathways[['JobProfile', 'JobFamily', 'similarity_score', 'career_move_type']].to_string(index=False))
    
    # 3. Workforce impact analysis
    workforce_impact = pd.read_sql_query("""
        SELECT 
            COUNT(*) as total_positions,
            COUNT(DISTINCT Division) as divisions_affected,
            COUNT(DISTINCT Location) as locations_affected,
            GROUP_CONCAT(DISTINCT Division) as division_list
        FROM positions 
        WHERE JobProfileID = ?
    """, conn, params=[job_profile_id])
    
    print(f"\n👥 Workforce Impact:")
    if not workforce_impact.empty:
        impact = workforce_impact.iloc[0]
        print(f"   • {impact['total_positions']} people in this role")
        print(f"   • {impact['divisions_affected']} divisions affected")
        print(f"   • {impact['locations_affected']} locations affected")
        print(f"   • Divisions: {impact['division_list']}")
    
    # 4. Determine story type based on data
    story_insights = determine_story_type(pathways, workforce_impact)
    
    return {
        'job_info': job_info,
        'pathways': pathways,
        'workforce_impact': workforce_impact.iloc[0] if not workforce_impact.empty else None,
        'story_insights': story_insights
    }
```

```python
def determine_story_type(pathways, workforce_impact):
    """
    Determine what type of story to tell based on the data patterns.
    This is the core branching logic for white paper generation.
    """
    
    if pathways.empty:
        return {
            'story_type': 'limited_options',
            'narrative': 'Limited direct pathways - focus on transferable skills',
            'tone': 'supportive',
            'focus': 'skills_transfer',
            'urgency': 'high_investment',
            'audience_focus': {
                'business_leaders': 'Skills investment and external recruitment may be needed',
                'hr_partners': 'Focus on comprehensive career counselling and support',
                'affected_colleagues': 'Explore transferable skills and alternative opportunities',
                'learning_teams': 'Design fundamental reskilling programs'
            }
        }
    
    # Analyze pathway quality
    high_similarity = pathways[pathways['similarity_score'] >= 0.7]
    medium_similarity = pathways[(pathways['similarity_score'] >= 0.4) & (pathways['similarity_score'] < 0.7)]
    
    avg_similarity = pathways['similarity_score'].mean()
    cross_family_moves = pathways[pathways['career_move_type'] == 'cross_family']
    lateral_moves = pathways[pathways['career_move_type'] == 'lateral']
    progression_moves = pathways[pathways['career_move_type'] == 'progression']
    
    # Determine story type based on opportunity quality
    if len(high_similarity) >= 3:
        story_type = 'excellent_opportunities'
        narrative = 'Multiple excellent transition opportunities available'
        tone = 'optimistic'
        focus = 'opportunity_rich'
        urgency = 'manageable_transition'
        
        audience_focus = {
            'business_leaders': 'Smooth transition expected with minimal business disruption',
            'hr_partners': 'Focus on matching preferences and facilitating smooth moves',
            'affected_colleagues': 'You have excellent options - let\'s explore what excites you most',
            'learning_teams': 'Minimal training required - focus on role-specific orientation'
        }
        
    elif len(high_similarity) >= 1 or len(medium_similarity) >= 3:
        story_type = 'good_opportunities'
        narrative = 'Strong transition opportunities with manageable development'
        tone = 'positive'
        focus = 'development_supported'
        urgency = 'planned_transition'
        
        audience_focus = {
            'business_leaders': 'Good transition prospects with some development investment',
            'hr_partners': 'Plan 3-6 month transition timeline with targeted development',
            'affected_colleagues': 'Strong opportunities available with focused skill development',
            'learning_teams': 'Design targeted programs for specific skill gaps'
        }
        
    elif len(medium_similarity) >= 1:
        story_type = 'development_opportunities'
        narrative = 'Strategic development opportunities requiring focused upskilling'
        tone = 'realistic'
        focus = 'skill_development'
        urgency = 'planned_development'
        
        audience_focus = {
            'business_leaders': 'Requires significant development investment but pathways exist',
            'hr_partners': 'Plan 6-12 month transition with comprehensive support',
            'affected_colleagues': 'Transformation opportunities that will expand your capabilities',
            'learning_teams': 'Design comprehensive development programs'
        }
        
    else:
        story_type = 'transformation_required'
        narrative = 'Significant transformation opportunities requiring strategic investment'
        tone = 'honest_supportive'
        focus = 'transformation'
        urgency = 'long_term_investment'
        
        audience_focus = {
            'business_leaders': 'Major reskilling investment needed or consider external recruitment',
            'hr_partners': 'Long-term career transition support and counselling required',
            'affected_colleagues': 'Major career pivot opportunities with comprehensive support',
            'learning_teams': 'Design extensive reskilling and career transformation programs'
        }
    
    # Add modifiers for special circumstances
    modifiers = []
    if len(cross_family_moves) > len(lateral_moves):
        modifiers.append('cross_functional_emphasis')
    if len(progression_moves) > len(lateral_moves):
        modifiers.append('career_advancement_focus')
    
    return {
        'story_type': story_type,
        'narrative': narrative,
        'tone': tone,
        'focus': focus,
        'urgency': urgency,
        'audience_focus': audience_focus,
        'modifiers': modifiers,
        'metrics': {
            'avg_similarity': round(avg_similarity, 3),
            'high_similarity_count': len(high_similarity),
            'medium_similarity_count': len(medium_similarity),
            'cross_family_count': len(cross_family_moves),
            'progression_count': len(progression_moves),
            'lateral_count': len(lateral_moves)
        }
    }

# Test with a sample job
test_job = sample_jobs.iloc[0]['JobProfileID']
analysis_result = analyze_role_for_story(test_job)
```

## 7. White Paper Content Generation

Now let's build the actual content generation system that takes our story insights and creates professional document content.


```python
def generate_white_paper_structure(analysis_data, scenario_type='role_sunsetting', target_audience='business_leaders'):
    """
    Generate a complete white paper structure based on analysis data and target scenario.
    This creates the framework that will be filled with actual content.
    """
    
    if not analysis_data:
        return None
    
    job_info = analysis_data['job_info']
    story_insights = analysis_data['story_insights']
    workforce_impact = analysis_data['workforce_impact']
    pathways = analysis_data['pathways']
    
    # Determine document title based on scenario
    scenario_titles = {
        'role_sunsetting': f"Career Transition Analysis: {job_info['JobProfile']} Role Sunset",
        'division_restructure': f"Workforce Redeployment Options: {job_info['JobProfile']} Division Restructure",
        'proactive_planning': f"Career Development Pathways: {job_info['JobProfile']} Strategic Planning",
        'skills_gap': f"Capability Building Analysis: {job_info['JobProfile']} Skills Development",
        'emergency_redeployment': f"Rapid Transition Options: {job_info['JobProfile']} Emergency Redeployment"
    }
    
    title = scenario_titles.get(scenario_type, f"Career Analysis: {job_info['JobProfile']}")
    
    # Build document structure
    structure = {
        'metadata': {
            'title': title,
            'subtitle': f"Strategic workforce transition analysis for Future Skills Team",
            'scenario': scenario_type,
            'target_audience': target_audience,
            'story_type': story_insights['story_type'],
            'urgency': story_insights['urgency'],
            'generated_date': datetime.now().strftime('%d %B %Y'),
            'confidence_level': 'High' if story_insights['metrics']['avg_similarity'] > 0.6 else 'Medium' if story_insights['metrics']['avg_similarity'] > 0.3 else 'Requires Validation'
        },
        
        'executive_summary': {
            'situation_overview': f"Analysis of transition opportunities for {workforce_impact['total_positions'] if workforce_impact else 'unknown number of'} colleagues currently in {job_info['JobProfile']} roles.",
            'key_finding': story_insights['narrative'],
            'confidence_statement': story_insights['audience_focus'][target_audience],
            'recommendation_summary': determine_primary_recommendation(story_insights, scenario_type),
            'impact_assessment': f"Affects {workforce_impact['divisions_affected'] if workforce_impact else 'multiple'} divisions across {workforce_impact['locations_affected'] if workforce_impact else 'multiple'} locations"
        },
        
        'sections': [
            {
                'section_number': '1',
                'title': 'Context & Current State',
                'content_type': 'descriptive',
                'purpose': 'Establish the situation and business context',
                'key_elements': [
                    f"Role Overview: {job_info['JobProfile']} in {job_info['JobFamily']} family",
                    f"Workforce Impact: {workforce_impact['total_positions'] if workforce_impact else 'TBD'} current positions",
                    f"Organisational Reach: {workforce_impact['division_list'] if workforce_impact else 'Multiple divisions'}",
                    f"Analysis Date: {datetime.now().strftime('%d %B %Y')}"
                ]
            },
            {
                'section_number': '2', 
                'title': 'Transition Opportunity Analysis',
                'content_type': 'analytical',
                'purpose': 'Present data-driven career pathway insights',
                'story_focus': story_insights['focus'],
                'tone': story_insights['tone'],
                'key_elements': generate_pathway_analysis_content(pathways, story_insights)
            },
            {
                'section_number': '3',
                'title': 'Skills Development Requirements', 
                'content_type': 'practical',
                'purpose': 'Identify specific skill gaps and development needs',
                'key_elements': generate_skills_development_content(analysis_data)
            },
            {
                'section_number': '4',
                'title': 'Implementation Roadmap',
                'content_type': 'strategic',
                'purpose': 'Provide actionable next steps and timeline',
                'urgency_level': story_insights['urgency'],
                'key_elements': generate_implementation_roadmap(story_insights, scenario_type)
            }
        ],
        
        'appendix': {
            'methodology': 'NAB Skills Similarity Engine - Pre-computed career pathway analysis',
            'data_sources': ['Job architecture', 'Skills taxonomy', 'Career pathways', 'Workforce positions'],
            'confidence_metrics': story_insights['metrics'],
            'detailed_pathways': pathways.to_dict('records') if not pathways.empty else []
        }
    }
    
    return structure

def determine_primary_recommendation(story_insights, scenario_type):
    """Generate the primary recommendation based on story type and scenario."""
    
    story_type = story_insights['story_type']
    urgency = story_insights['urgency']
    
    recommendations = {
        'excellent_opportunities': {
            'role_sunsetting': 'Proceed with confidence - multiple excellent transition options identified',
            'division_restructure': 'Smooth redeployment expected with minimal disruption',
            'proactive_planning': 'Strong internal career progression opportunities available'
        },
        'good_opportunities': {
            'role_sunsetting': 'Solid transition options with targeted development support',
            'division_restructure': 'Planned redeployment with 3-6 month development timeline',
            'proactive_planning': 'Good career advancement potential with focused skill building'
        },
        'development_opportunities': {
            'role_sunsetting': 'Viable transitions requiring strategic development investment',
            'division_restructure': 'Redeployment possible with comprehensive development programs',
            'proactive_planning': 'Transformation opportunities requiring 6-12 month development'
        },
        'transformation_required': {
            'role_sunsetting': 'Major reskilling required - consider external recruitment for urgent needs',
            'division_restructure': 'Significant redeployment challenges - long-term planning required',
            'proactive_planning': 'Fundamental career transformation opportunities with extensive support'
        }
    }
    
    return recommendations.get(story_type, {}).get(scenario_type, 'Strategic analysis required for transition planning')

def generate_pathway_analysis_content(pathways, story_insights):
    """Generate specific content for the pathway analysis section."""
    
    if pathways.empty:
        return [
            "Limited direct career pathways identified in current analysis",
            "Focus on transferable skills and alternative career directions", 
            "Consider external market opportunities and comprehensive reskilling"
        ]
    
    top_pathways = pathways.head(5)
    
    content = [
        f"Analysis identified {len(pathways)} viable career pathways",
        f"Average similarity score: {story_insights['metrics']['avg_similarity']} (scale 0-1)",
        f"Top career opportunities:"
    ]
    
    for i, (_, pathway) in enumerate(top_pathways.iterrows(), 1):
        content.append(f"   {i}. {pathway['JobProfile']} - Similarity: {pathway['similarity_score']:.2f}")
    
    # Add movement type analysis
    if story_insights['metrics']['cross_family_count'] > 0:
        content.append(f"Cross-functional opportunities: {story_insights['metrics']['cross_family_count']} pathways")
    
    if story_insights['metrics']['progression_count'] > 0:
        content.append(f"Career advancement opportunities: {story_insights['metrics']['progression_count']} pathways")
    
    return content

def generate_skills_development_content(analysis_data):
    """Generate content for skills development section."""
    
    pathways = analysis_data['pathways']
    
    if pathways.empty:
        return [
            "Comprehensive skills assessment required",
            "Focus on identifying transferable competencies",
            "Design fundamental reskilling program"
        ]
    
    # Analyze skill requirements from top pathways
    avg_shared_skills = pathways['shared_skills_count'].mean()
    max_shared_skills = pathways['shared_skills_count'].max()
    min_shared_skills = pathways['shared_skills_count'].min()
    
    content = [
        f"Skills overlap analysis across {len(pathways)} pathways",
        f"Average shared skills: {avg_shared_skills:.1f} skills",
        f"Range: {min_shared_skills}-{max_shared_skills} skills overlap",
        "Development priority areas:"
    ]
    
    # Add specific development recommendations based on similarity scores
    high_similarity_paths = pathways[pathways['similarity_score'] >= 0.7]
    medium_similarity_paths = pathways[(pathways['similarity_score'] >= 0.4) & (pathways['similarity_score'] < 0.7)]
    
    if not high_similarity_paths.empty:
        content.append(f"• Immediate transition readiness: {len(high_similarity_paths)} roles")
        content.append("• Focus: Role-specific orientation and process training")
    
    if not medium_similarity_paths.empty:
        content.append(f"• Medium-term development: {len(medium_similarity_paths)} roles")  
        content.append("• Focus: Targeted skill gap training (3-6 months)")
    
    return content

def generate_implementation_roadmap(story_insights, scenario_type):
    """Generate implementation roadmap based on urgency and story type."""
    
    urgency = story_insights['urgency']
    story_type = story_insights['story_type']
    
    roadmaps = {
        'manageable_transition': [
            "Phase 1 (Weeks 1-2): Individual preference discussions and role matching",
            "Phase 2 (Weeks 3-4): Specific role assignments and handover planning", 
            "Phase 3 (Weeks 5-8): Role transition and orientation",
            "Success criteria: Smooth transition with minimal productivity loss"
        ],
        'planned_transition': [
            "Phase 1 (Month 1): Skills assessment and development planning",
            "Phase 2 (Months 2-4): Targeted development and preparation",
            "Phase 3 (Months 5-6): Role transition and support",
            "Success criteria: Successful role transition with development investment"
        ],
        'planned_development': [
            "Phase 1 (Months 1-2): Comprehensive skills gap analysis", 
            "Phase 2 (Months 3-8): Structured development program",
            "Phase 3 (Months 9-12): Gradual role transition with mentoring",
            "Success criteria: Successful transformation with comprehensive support"
        ],
        'long_term_investment': [
            "Phase 1 (Months 1-3): Strategic career counselling and pathway planning",
            "Phase 2 (Months 4-12): Extensive reskilling and development",
            "Phase 3 (Months 13-18): Gradual transition with ongoing support",
            "Success criteria: Fundamental career transformation achieved"
        ]
    }
    
    return roadmaps.get(urgency, [
        "Phase 1: Detailed individual assessment required",
        "Phase 2: Customised development planning", 
        "Phase 3: Structured transition support",
        "Success criteria: To be determined based on individual circumstances"
    ])

# Test the structure generation
if analysis_result:
    white_paper_structure = generate_white_paper_structure(analysis_result, 'role_sunsetting', 'business_leaders')
    
    print("\n=== WHITE PAPER STRUCTURE GENERATED ===")
    print(f"📄 Title: {white_paper_structure['metadata']['title']}")
    print(f"🎯 Story Type: {white_paper_structure['metadata']['story_type']}")
    print(f"⏰ Urgency: {white_paper_structure['metadata']['urgency']}")
    print(f"🔍 Confidence: {white_paper_structure['metadata']['confidence_level']}")
    
    print(f"\n📋 Executive Summary:")
    exec_summary = white_paper_structure['executive_summary']
    for key, value in exec_summary.items():
        print(f"   • {key.replace('_', ' ').title()}: {value}")
    
    print(f"\n📝 Document Sections:")
    for section in white_paper_structure['sections']:
        print(f"   {section['section_number']}. {section['title']} ({section['content_type']})")
        print(f"      Purpose: {section['purpose']}")
        if 'key_elements' in section and section['key_elements']:
            print(f"      Elements: {len(section['key_elements'])} key points")
```

```python
def generate_sample_executive_summary(white_paper_structure):
    """Generate a complete executive summary using our structure."""
    
    if not white_paper_structure:
        return "Unable to generate executive summary - no structure provided"
    
    metadata = white_paper_structure['metadata']
    exec_summary = white_paper_structure['executive_summary']
    
    # Format as professional document content
    content = f"""
# Executive Summary

## {metadata['title']}

**Prepared for**: Future Skills Team  
**Analysis Date**: {metadata['generated_date']}  
**Confidence Level**: {metadata['confidence_level']}  
**Scenario Type**: {metadata['scenario'].replace('_', ' ').title()}

---

### Situation Overview

{exec_summary['situation_overview']}

### Key Finding

{exec_summary['key_finding']}

### Strategic Recommendation

{exec_summary['recommendation_summary']}

### Implementation Confidence

{exec_summary['confidence_statement']}

### Impact Assessment

{exec_summary['impact_assessment']}

---

**Document Status**: {metadata['story_type'].replace('_', ' ').title()} | **Urgency**: {metadata['urgency'].replace('_', ' ').title()}

*This analysis was generated using the NAB Skills Intelligence Platform and is intended for strategic workforce planning purposes.*
    """
    
    return content.strip()

def generate_scenario_comparison():
    """Generate white papers for different scenarios to show variation."""
    
    if not analysis_result:
        print("❌ No analysis result available for comparison")
        return
    
    scenarios = [
        ('role_sunsetting', 'business_leaders'),
        ('division_restructure', 'hr_partners'), 
        ('proactive_planning', 'affected_colleagues'),
        ('emergency_redeployment', 'business_leaders')
    ]
    
    print("=== SCENARIO COMPARISON ===")
    
    for scenario_type, audience in scenarios:
        print(f"\n🎯 Scenario: {scenario_type.replace('_', ' ').title()} | Audience: {audience.replace('_', ' ').title()}")
        
        structure = generate_white_paper_structure(analysis_result, scenario_type, audience)
        
        if structure:
            print(f"   📄 Title: {structure['metadata']['title']}")
            print(f"   💡 Key Message: {structure['executive_summary']['key_finding']}")
            print(f"   🎯 Recommendation: {structure['executive_summary']['recommendation_summary']}")
            print(f"   ⏰ Urgency: {structure['metadata']['urgency']}")
        else:
            print("   ❌ Failed to generate structure")

# Test content generation if we have results
if analysis_result:
    # Generate structure
    sample_structure = generate_white_paper_structure(analysis_result, 'role_sunsetting', 'business_leaders')
    
    if sample_structure:
        # Generate and display executive summary
        exec_summary_content = generate_sample_executive_summary(sample_structure)
        print("=== SAMPLE EXECUTIVE SUMMARY ===")
        print(exec_summary_content)
        
        # Show scenario variations
        generate_scenario_comparison()
    else:
        print("❌ Unable to generate white paper structure")
else:
    print("⚠️ No analysis result available - need to run data analysis first")
```
