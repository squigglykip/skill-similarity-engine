```python
import pandas as pd

# Update these paths to your actual output files
file_unweighted = r'C:\Users\kipjo\OneDrive\Documents\GitHub\skill-similarity-engine\data\poc\output\poc_run_20250528_162415\job_asymmetric_coverage.csv'
file_weighted = r'C:\Users\kipjo\OneDrive\Documents\GitHub\skill-similarity-engine\data\poc\output\poc_run_20250528_191717\job_asymmetric_coverage.csv'

df_unweighted = pd.read_csv(file_unweighted)
df_weighted = pd.read_csv(file_weighted)
```

```python
print("Displaying the first few rows of the unweighted dataframe:")
print(df_unweighted.head())
print("\n")  # Adding space between outputs

print("Displaying the first few rows of the weighted dataframe:")
print(df_weighted.head())
print("\n")  # Adding space between outputs
```

```python
# Merge on job_from and job_to
df_compare = df_unweighted.merge(
    df_weighted,
    on=['job_from', 'job_to'],
    suffixes=('_unweighted', '_weighted')
)
```

```python
# Calculate the difference
df_compare['similarity_diff'] = df_compare['similarity_weighted'] - df_compare['similarity_unweighted']

# Show summary statistics
print(df_compare['similarity_diff'].describe())

# Show a few of the largest changes
df_compare_sorted = df_compare.reindex(df_compare.similarity_diff.abs().sort_values(ascending=False).index)
df_compare_sorted[['job_from', 'job_to', 'similarity_unweighted', 'similarity_weighted', 'similarity_diff']].head(10)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,4))
plt.hist(df_compare['similarity_diff'], bins=50, color='purple', edgecolor='black', alpha=0.7)
plt.title('Distribution of Change in Similarity (Weighted - Unweighted)')
plt.xlabel('Difference in Similarity')
plt.ylabel('Number of Job Pairs')
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6,6))
plt.scatter(
    df_compare['similarity_unweighted'],
    df_compare['similarity_weighted'],
    alpha=0.3, s=10, color='teal'
)
plt.plot([0,1],[0,1],'--',color='grey')
plt.xlabel('Unweighted Similarity')
plt.ylabel('Weighted Similarity')
plt.title('Weighted vs Unweighted Similarity')
plt.grid(True, alpha=0.4)
plt.tight_layout()
plt.show()
```
