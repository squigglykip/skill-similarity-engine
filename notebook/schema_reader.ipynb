```python
import os
import pandas as pd

def print_csv_schemas(folder_path):
    """
    Prints the schema (field names and data types) of all CSV files in the specified folder.

    :param folder_path: Path to the folder containing CSV files.
    """
    # List all files in the given folder
    files = os.listdir(folder_path)
    
    # Filter out non-CSV files
    csv_files = [file for file in files if file.endswith('.csv')]
    
    # Process each CSV file
    for csv_file in csv_files:
        file_path = os.path.join(folder_path, csv_file)
        try:
            # Read the CSV file into a DataFrame
            df = pd.read_csv(file_path)
            
            # Print the schema
            print(f"Schema for {csv_file}:")
            for column in df.columns:
                print(f"  - {column}: {df[column].dtype}")
            print("\n")
        except Exception as e:
            print(f"Error reading {csv_file}: {e}")

# Example usage (as a comment):
print_csv_schemas(r"C:\Users\kipjo\OneDrive\Documents\GitHub\skill-similarity-engine\data\poc\colleague-services-v2")
```
