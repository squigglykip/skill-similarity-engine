```python
import pandas as pd

# File path for engine output
engine_output_path = r'C:\Users\kipjo\OneDrive\Documents\GitHub\skill-similarity-engine\data\poc\output\poc_run_20250529_170826\job_asymmetric_coverage.csv'

# Load engine output
df_engine = pd.read_csv(engine_output_path)

# Job pair to inspect
job1_composite = 'R0323_55026912'
job2_composite = 'R0350_55033472'

# Filter for the specific job pair
row = df_engine[
    (df_engine['job_from'] == job1_composite) &
    (df_engine['job_to'] == job2_composite)
]

if not row.empty:
    similarity_score = row.iloc[0]['similarity']
    print(f"Engine similarity score from {job1_composite} to {job2_composite}: {similarity_score:.4f}")
else:
    print("No similarity score found for this job pair in the engine output.")
```

```python
# File paths for raw data
job_data_path = r'C:\Users\kipjo\OneDrive\Documents\GitHub\skill-similarity-engine\data\poc\colleague-services-v2\job_data.csv'
job_skill_mapping_path = r'C:\Users\kipjo\OneDrive\Documents\GitHub\skill-similarity-engine\data\poc\colleague-services-v2\job_skill_mapping.csv'
skill_data_path = r'C:\Users\kipjo\OneDrive\Documents\GitHub\skill-similarity-engine\data\poc\colleague-services-v2\skill_data.csv'

# Load raw data
df_jobs = pd.read_csv(job_data_path)
df_job_skill = pd.read_csv(job_skill_mapping_path)
df_skills = pd.read_csv(skill_data_path)

# Create composite job IDs
df_jobs['unique_job_id'] = df_jobs['JobID'].astype(str) + '_' + df_jobs['Org Unit Number'].astype(str)

# Merge composite job ID into job-skill mapping
df_job_skill = df_job_skill.merge(
    df_jobs[['JobID', 'Org Unit Number', 'unique_job_id']],
    on='JobID',
    how='left'
)

# Remove rows where Skill_ID is nan, empty, or the string 'nan'
df_job_skill = df_job_skill[
    df_job_skill['Skill_ID'].notna() &
    (df_job_skill['Skill_ID'] != '') &
    (df_job_skill['Skill_ID'].astype(str).str.lower() != 'nan')
]

# Function to get skills for a composite job ID
def get_skills_for_composite(composite_id):
    return set(df_job_skill[df_job_skill['unique_job_id'] == composite_id]['Skill_ID'])

# Get skill sets for each job
job1_skills = get_skills_for_composite(job1_composite)
job2_skills = get_skills_for_composite(job2_composite)

print(f"Job 1 ({job1_composite}) has {len(job1_skills)} skills.")
print(f"Job 2 ({job2_composite}) has {len(job2_skills)} skills.")
print(f"\nSample skills for Job 1: {list(job1_skills)[:5]}")
print(f"Sample skills for Job 2: {list(job2_skills)[:5]}")
```

```python
# Get sets of composite job IDs from both sources
engine_job_ids = set(df_engine['job_from']).union(set(df_engine['job_to']))
raw_job_ids = set(df_job_skill['unique_job_id'])

# Find intersection
common_job_ids = engine_job_ids & raw_job_ids
print(f"Number of common job IDs: {len(common_job_ids)}")
print("Sample common job IDs:", list(common_job_ids)[:10])

# Find a job pair in the engine output where both jobs exist in the raw data
for idx, row in df_engine.iterrows():
    if row['job_from'] in raw_job_ids and row['job_to'] in raw_job_ids:
        print("Example job pair found:")
        print("job_from:", row['job_from'])
        print("job_to:", row['job_to'])
        break
```

```python
# Calculate the overlap: percentage of job1's skills found in job2
if job1_skills:
    overlap_count = len(job1_skills & job2_skills)
    manual_coverage = overlap_count / len(job1_skills)
    print(f"\nManual calculation: {overlap_count} of {len(job1_skills)} Job 1 skills are also in Job 2.")
    print(f"Manual coverage (Job 1 in Job 2): {manual_coverage:.4f}")
else:
    print("No skills found for Job 1.")

# Show both values side by side if available
if 'similarity_score' in locals():
    print(f"\nEngine similarity score: {similarity_score:.4f}")
    print(f"Manual skill overlap:   {manual_coverage:.4f}")
    print(f"Difference:            {abs(similarity_score - manual_coverage):.6f}")
```
