# Workforce Context Data Analysis

Analysis of the dummy workforce context data generated for Phase 8.1 webapp development.

This notebook explores the structure, quality, and relationships in our dummy SAP/HRIS data to validate it's suitable for webapp prototyping.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Set display options
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
plt.style.use('default')
sns.set_palette("husl")

# Load the dummy workforce context data
data_path = Path("../data/workforce_context/dummy_workforce_context.csv")
print(f"Loading data from: {data_path}")
print(f"File exists: {data_path.exists()}")

if data_path.exists():
    df = pd.read_csv(data_path)
    print(f"\nLoaded {len(df):,} records with {len(df.columns)} columns")
else:
    print("⚠️  Data file not found. Run the generate_dummy_workforce_context.py script first.")
```

## Data Schema Overview

Let's examine the structure and schema of our dummy workforce context data.


```python
# Schema information
if 'df' in locals():
    print("=== SCHEMA INFORMATION ===")
    print(f"Shape: {df.shape}")
    print(f"Columns: {len(df.columns)}")
    print("\nColumn List:")
    for i, col in enumerate(df.columns, 1):
        print(f"{i:2d}. {col}")
    
    print("\n=== DATA TYPES ===")
    print(df.dtypes)
    
    print("\n=== SAMPLE DATA ===")
    display(df.head())
```

## Data Quality Assessment

Check for missing values, data completeness, and basic validation.


```python
# Data quality analysis
if 'df' in locals():
    print("=== MISSING VALUES ANALYSIS ===")
    missing_stats = df.isnull().sum()
    missing_pct = (missing_stats / len(df) * 100).round(1)
    missing_df = pd.DataFrame({
        'Missing Count': missing_stats,
        'Missing %': missing_pct
    }).sort_values('Missing Count', ascending=False)
    
    print(f"Columns with missing data:")
    print(missing_df[missing_df['Missing Count'] > 0])
    
    print(f"\n=== POSITION vs EMPLOYEE FILL RATES ===")
    total_positions = len(df)
    filled_positions = df['Employee Number'].notna().sum()
    vacant_positions = total_positions - filled_positions
    
    print(f"Total Positions: {total_positions:,}")
    print(f"Filled Positions: {filled_positions:,} ({filled_positions/total_positions*100:.1f}%)")
    print(f"Vacant Positions: {vacant_positions:,} ({vacant_positions/total_positions*100:.1f}%)")
    
    print(f"\n=== UNIQUE VALUE COUNTS ===")
    key_fields = ['Position Number', 'Employee Number', 'Location', 'Salary Group', 'People Leader Flag']
    for field in key_fields:
        if field in df.columns:
            unique_count = df[field].nunique()
            print(f"{field}: {unique_count:,} unique values")
    
    # Check for duplicates
    print(f"\n=== DUPLICATE CHECK ===")
    position_dups = df['Position Number'].duplicated().sum()
    employee_dups = df['Employee Number'].dropna().duplicated().sum()
    print(f"Duplicate Position Numbers: {position_dups}")
    print(f"Duplicate Employee Numbers: {employee_dups}")
```

## Organizational Hierarchy Analysis

Explore the 10-level organizational structure to understand business unit distribution.


```python
# Organizational hierarchy analysis
if 'df' in locals():
    print("=== ORGANIZATIONAL HIERARCHY LEVELS ===")
    
    # Check each org level
    org_levels = [f'ORG UNIT NAME_{i}' for i in range(1, 11)]
    
    for level_col in org_levels:
        if level_col in df.columns:
            level_num = level_col.split('_')[-1]
            unique_units = df[level_col].nunique()
            sample_units = df[level_col].value_counts().head(5)
            
            print(f"\n--- Level {level_num} ({level_col}) ---")
            print(f"Unique units: {unique_units}")
            print("Top 5 units:")
            for unit, count in sample_units.items():
                print(f"  {unit}: {count:,} positions")
    
    print(f"\n=== BUSINESS UNIT DISTRIBUTION ===")
    # Level 3 should be business units
    if 'ORG UNIT NAME_3' in df.columns:
        bu_dist = df['ORG UNIT NAME_3'].value_counts()
        print("Business Unit distribution:")
        for bu, count in bu_dist.items():
            print(f"  {bu}: {count:,} positions ({count/len(df)*100:.1f}%)")
    
    print(f"\n=== LOCATION DISTRIBUTION ===")
    if 'Location' in df.columns:
        location_dist = df['Location'].value_counts()
        print("Location distribution:")
        for location, count in location_dist.items():
            print(f"  {location}: {count:,} positions ({count/len(df)*100:.1f}%)")
```

## Position and Employee Analysis

Analyze position types, salary groups, and people leader relationships.


```python
# Position and employee analysis
if 'df' in locals():
    print("=== POSITION NAME ANALYSIS ===")
    if 'Position Name' in df.columns:
        # Extract position level patterns
        position_sample = df['Position Name'].value_counts().head(10)
        print("Top 10 position names:")
        for pos, count in position_sample.items():
            print(f"  {pos}: {count}")
        
        # Analyze position level words
        all_positions = df['Position Name'].dropna()
        level_words = ['Analyst', 'Manager', 'Senior', 'Lead', 'Director', 'Principal', 'Associate']
        level_counts = {}
        for word in level_words:
            count = all_positions.str.contains(word, case=False).sum()
            level_counts[word] = count
        
        print(f"\nPosition level distribution:")
        for word, count in sorted(level_counts.items(), key=lambda x: x[1], reverse=True):
            print(f"  {word}: {count:,} positions")
    
    print(f"\n=== SALARY GROUP ANALYSIS ===")
    if 'Salary Group' in df.columns:
        salary_dist = df['Salary Group'].value_counts()
        print("Salary group distribution:")
        for group, count in salary_dist.items():
            print(f"  {group}: {count:,} positions ({count/len(df)*100:.1f}%)")
    
    print(f"\n=== PEOPLE LEADER ANALYSIS ===")
    if 'People Leader Flag' in df.columns:
        leader_dist = df['People Leader Flag'].value_counts()
        print("People leader distribution:")
        for flag, count in leader_dist.items():
            print(f"  {flag}: {count:,} positions ({count/len(df)*100:.1f}%)")
    
    print(f"\n=== EMPLOYEE GROUP ANALYSIS ===")
    if 'Employee Group' in df.columns:
        emp_group_dist = df['Employee Group'].value_counts()
        print("Employee group distribution:")
        for group, count in emp_group_dist.items():
            print(f"  {group}: {count:,} employees")
    
    print(f"\n=== FTE ANALYSIS ===")
    if 'FTE (raw value in SAP)' in df.columns:
        fte_dist = df['FTE (raw value in SAP)'].value_counts().sort_index()
        print("FTE distribution:")
        for fte, count in fte_dist.items():
            print(f"  {fte}: {count:,} positions")
```

## Webapp Query Simulation

Test the types of filtering and querying the webapp will need to perform.


```python
# Simulate webapp queries
if 'df' in locals():
    print("=== WEBAPP QUERY SIMULATIONS ===")
    print("These simulate the types of filtering users will do in the webapp.\n")
    
    # Query 1: Technology positions in Sydney
    print("Query 1: Technology positions in Sydney")
    tech_sydney = df[
        (df['ORG UNIT NAME_3'].str.contains('Technology', case=False, na=False)) &
        (df['Location'] == 'Sydney')
    ]
    print(f"Results: {len(tech_sydney):,} positions")
    if len(tech_sydney) > 0:
        salary_breakdown = tech_sydney['Salary Group'].value_counts()
        print("Salary group breakdown:")
        for group, count in salary_breakdown.items():
            print(f"  {group}: {count}")
    
    # Query 2: Senior roles (Group 5+) with people leader responsibility
    print(f"\nQuery 2: Senior roles (Group 5+) with people leader responsibility")
    senior_leaders = df[
        (df['Salary Group'].isin(['Group 5', 'Group 6', 'Group 7'])) &
        (df['People Leader Flag'] == 'People Leader')
    ]
    print(f"Results: {len(senior_leaders):,} positions")
    if len(senior_leaders) > 0:
        location_breakdown = senior_leaders['Location'].value_counts()
        print("Location breakdown:")
        for location, count in location_breakdown.head().items():
            print(f"  {location}: {count}")
    
    # Query 3: All analysts across business units
    print(f"\nQuery 3: All analyst positions across business units")
    analysts = df[df['Position Name'].str.contains('Analyst', case=False, na=False)]
    print(f"Results: {len(analysts):,} positions")
    if len(analysts) > 0:
        bu_breakdown = analysts['ORG UNIT NAME_3'].value_counts()
        print("Business unit breakdown:")
        for bu, count in bu_breakdown.head().items():
            print(f"  {bu}: {count}")
    
    # Query 4: Vacant positions by business unit
    print(f"\nQuery 4: Vacant positions by business unit")
    vacant = df[df['Employee Number'].isna()]
    print(f"Results: {len(vacant):,} vacant positions")
    if len(vacant) > 0:
        vacant_by_bu = vacant['ORG UNIT NAME_3'].value_counts()
        print("Vacant positions by business unit:")
        for bu, count in vacant_by_bu.head().items():
            print(f"  {bu}: {count}")
    
    # Query 5: People leaders and their span of control
    print(f"\nQuery 5: People leader span of control analysis")
    leaders = df[df['People Leader Number'].notna()]
    if len(leaders) > 0:
        span_counts = leaders['People Leader Number'].value_counts()
        print(f"Total people leaders: {span_counts.nunique():,}")
        print(f"Average span of control: {span_counts.mean():.1f}")
        print(f"Max span of control: {span_counts.max()}")
        print(f"Leaders with 5+ direct reports: {(span_counts >= 5).sum()}")
```

## Webapp Dropdown Data Preparation

Generate the data structures needed for webapp dropdowns and filters.


```python
# Generate webapp dropdown data
if 'df' in locals():
    print("=== WEBAPP DROPDOWN OPTIONS ===")
    print("These are the filter options we could offer users in the webapp.\n")
    
    # Business Units (for filtering similar roles)
    business_units = sorted(df['ORG UNIT NAME_3'].dropna().unique())
    print(f"Business Units ({len(business_units)} options):")
    for bu in business_units:
        count = (df['ORG UNIT NAME_3'] == bu).sum()
        print(f"  {bu} ({count:,} positions)")
    
    # Locations (for geographic filtering)
    print(f"\nLocations ({df['Location'].nunique()} options):")
    locations = df['Location'].value_counts()
    for location, count in locations.items():
        print(f"  {location} ({count:,} positions)")
    
    # Salary Groups (for career level filtering)
    print(f"\nSalary Groups ({df['Salary Group'].nunique()} options):")
    salary_groups = df['Salary Group'].value_counts()
    for group, count in salary_groups.items():
        print(f"  {group} ({count:,} positions)")
    
    # People Leader Flag (for leadership filtering)
    print(f"\nLeadership Roles:")
    leader_flags = df['People Leader Flag'].value_counts()
    for flag, count in leader_flags.items():
        print(f"  {flag} ({count:,} positions)")
    
    # Position Level Analysis (derived from position names)
    print(f"\nPosition Levels (derived from position names):")
    level_keywords = {
        'Executive': ['Executive', 'Chief', 'EVP'],
        'Director': ['Director'],
        'Principal': ['Principal'],
        'Senior': ['Senior'],
        'Lead': ['Lead'],
        'Manager': ['Manager'],
        'Associate': ['Associate'],
        'Analyst': ['Analyst'],
        'Specialist': ['Specialist'],
        'Consultant': ['Consultant'],
        'Developer': ['Developer'],
        'Engineer': ['Engineer']
    }
    
    position_levels = {}
    for level, keywords in level_keywords.items():
        count = 0
        for keyword in keywords:
            count += df['Position Name'].str.contains(keyword, case=False, na=False).sum()
        if count > 0:
            position_levels[level] = count
    
    for level, count in sorted(position_levels.items(), key=lambda x: x[1], reverse=True):
        print(f"  {level}: {count:,} positions")
    
    print(f"\n=== SUMMARY FOR WEBAPP DESIGN ===")
    print(f"Recommended dropdown filters:")
    print(f"• Business Unit: {len(business_units)} options")
    print(f"• Location: {df['Location'].nunique()} options") 
    print(f"• Salary Group: {df['Salary Group'].nunique()} options")
    print(f"• Leadership Role: {df['People Leader Flag'].nunique()} options")
    print(f"• Employment Type: {df['Employee Group'].nunique()} options (for filled positions)")
    print(f"\nThis provides rich filtering for {len(df):,} total positions.")
```

## Data Export for Development

Export key datasets for webapp development and testing.


```python
# Export key datasets for webapp development
if 'df' in locals():
    print("=== EXPORTING DEVELOPMENT DATASETS ===")
    
    export_dir = Path("../data/workforce_context/")
    
    # 1. Business unit mapping for dropdowns
    business_unit_mapping = df.groupby('ORG UNIT NAME_3').agg({
        'Position Number': 'count',
        'Employee Number': lambda x: x.notna().sum(),  # Count filled positions
        'Location': lambda x: x.mode().iloc[0] if len(x.mode()) > 0 else 'Unknown'  # Most common location
    }).reset_index()
    business_unit_mapping.columns = ['Business_Unit', 'Total_Positions', 'Filled_Positions', 'Primary_Location']
    business_unit_mapping['Vacancy_Rate'] = ((business_unit_mapping['Total_Positions'] - business_unit_mapping['Filled_Positions']) / business_unit_mapping['Total_Positions'] * 100).round(1)
    
    bu_file = export_dir / "business_unit_summary.csv"
    business_unit_mapping.to_csv(bu_file, index=False)
    print(f"✅ Exported business unit summary: {bu_file}")
    
    # 2. Location summary for geographic filtering
    location_summary = df.groupby(['Location', 'Rg']).agg({
        'Position Number': 'count',
        'Employee Number': lambda x: x.notna().sum(),
        'ORG UNIT NAME_3': lambda x: x.nunique()  # Number of different business units
    }).reset_index()
    location_summary.columns = ['Location', 'Region', 'Total_Positions', 'Filled_Positions', 'Business_Units']
    
    location_file = export_dir / "location_summary.csv"
    location_summary.to_csv(location_file, index=False)
    print(f"✅ Exported location summary: {location_file}")
    
    # 3. Salary group analysis for career level filtering
    salary_summary = df.groupby('Salary Group').agg({
        'Position Number': 'count',
        'Employee Number': lambda x: x.notna().sum(),
        'People Leader Flag': lambda x: (x == 'People Leader').sum()
    }).reset_index()
    salary_summary.columns = ['Salary_Group', 'Total_Positions', 'Filled_Positions', 'People_Leaders']
    
    salary_file = export_dir / "salary_group_summary.csv"
    salary_summary.to_csv(salary_file, index=False)
    print(f"✅ Exported salary group summary: {salary_file}")
    
    # 4. Sample positions for testing (100 filled positions with full context)
    sample_positions = df[df['Employee Number'].notna()].sample(min(100, len(df))).copy()
    sample_file = export_dir / "sample_positions.csv"
    # sample_positions.to_csv(sample_file, index=False)
    print(f"✅ Exported sample positions for testing: {sample_file}")
    
    print(f"\n=== VALIDATION SUMMARY ===")
    print(f"Total workforce context records: {len(df):,}")
    print(f"Positions filled: {df['Employee Number'].notna().sum():,}")
    print(f"Vacant positions: {df['Employee Number'].isna().sum():,}")
    print(f"Business units: {df['ORG UNIT NAME_3'].nunique()}")
    print(f"Locations: {df['Location'].nunique()}")
    print(f"Salary groups: {df['Salary Group'].nunique()}")
    print(f"People leaders: {(df['People Leader Flag'] == 'People Leader').sum():,}")
    print(f"\n✅ Dummy workforce context data is ready for webapp development!")
```
